```python
import sqlite3

# Connect to a SQLite database (this creates 'Project.db' if it doesn't exist)
# You can use ':memory:' for an in-memory database that disappears after the script runs.
conn = sqlite3.connect('Project.db')
cursor = conn.cursor()

# Create the ORDERS table
# INTEGER, REAL, TEXT are SQLite equivalents for int, double, varchar
cursor.execute('''
CREATE TABLE ORDERS (
    order_id INTEGER,
    order_time REAL,
    customer_id TEXT,
    customer_name TEXT,
    address_pincode INTEGER,
    apartment_floor INTEGER
);
''')

# --- Example: Insert some data for demonstration ---
cursor.execute("INSERT INTO ORDERS VALUES (1, 12.30, 'C1', 'Alice', 12345, 10)")
cursor.execute("INSERT INTO ORDERS VALUES (2, NULL, 'C2', 'Bob', 54321, 5)")
cursor.execute("INSERT INTO ORDERS VALUES (3, 13.15, 'C3', 'Charlie', 67890, 12)")
cursor.execute("INSERT INTO ORDERS VALUES (4, 12.50, 'C4', 'David', 11223, 8)")
cursor.execute("INSERT INTO ORDERS VALUES (5, 14.00, 'C5', 'Anna', 98765, 7)")

print("Database 'Project.db' created and sample data inserted.\n")

# --- Execute the original SELECT queries ---

# Query 1: SELECT order_id FROM Orders WHERE order_time IS NULL OR order_time="";
# Note: In SQLite, an empty string might be treated differently than NULL for REAL types.
# We'll check for IS NULL directly.
print("Query 1: order_id from Orders where order_time IS NULL or empty string:")
cursor.execute("SELECT order_id FROM ORDERS WHERE order_time IS NULL")
results1 = cursor.fetchall()
print(results1)

print("\nQuery 2: All orders where order_time is between 12 and 13 (inclusive):")
cursor.execute("SELECT * FROM ORDERS WHERE order_time >= 12 AND order_time <= 13")
results2 = cursor.fetchall()
print(results2)

print("\nQuery 3: All orders ordered by apartment_floor:")
cursor.execute("SELECT * FROM ORDERS ORDER BY apartment_floor")
results3 = cursor.fetchall()
print(results3)

print("\nQuery 4: Distinct customer_id from Orders where customer_name LIKE '%a':")
cursor.execute("SELECT DISTINCT customer_id FROM ORDERS WHERE customer_name LIKE '%a'")
results4 = cursor.fetchall()
print(results4)

# Commit any changes (like data insertion) and close the connection
conn.commit()
conn.close()

print("\nOperations completed and database connection closed.")
```
